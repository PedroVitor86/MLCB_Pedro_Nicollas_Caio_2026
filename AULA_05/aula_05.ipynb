```python
import re
import numpy as np
import pandas as pd
import nltk
import spacy
from nltk.corpus import stopwords
from gensim.models import FastText
from sklearn.linear_model import LogisticRegression
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import (
    accuracy_score, precision_score, recall_score, f1_score,
    classification_report
)

nltk.download("stopwords", quiet=True)
stop_words_pt = set(stopwords.words("portuguese"))
nlp = spacy.load("pt_core_news_sm")

def limpar_e_lemmatizar(texto):
    """
    Realiza a limpeza, remoção de acentos/símbolos, tokenização,
    remoção de stopwords e lemmatization do texto.
    """
    texto_limpo = texto.lower()
    texto_limpo = re.sub(r'[^a-záàâãéèêíïóôõöúçñ\s]', '', texto_limpo)

    doc = nlp(texto_limpo)
    tokens_filtrados = []

    for token in doc:
        if token.is_space:
            continue
        if token.text in stop_words_pt:
            continue
        if len(token.text) <= 1:
            continue
        tokens_filtrados.append(token.lemma_)

    return " ".join(tokens_filtrados)

def analisar_preprocessamento(texto):
    """
    Função de diagnóstico que compara o texto original com a versão processada.
    """
    texto_limpo_base = re.sub(r'[^a-záàâãéèêíïóôõöúçñ\s]', '', texto.lower())
    tokens_antes = texto_limpo_base.split()

    texto_normalizado = limpar_e_lemmatizar(texto)
    tokens_depois = texto_normalizado.split()
    tokens_removidos = [t for t in tokens_antes if t not in tokens_depois]

    print("=" * 50)
    print("DIAGNÓSTICO DE PRÉ-PROCESSAMENTO")
    print("=" * 50)
    print(f"Texto original:\n{texto}\n")
    print(f"Texto normalizado:\n{texto_normalizado}\n")
    print(f"Quantidade de caracteres: {len(texto)}")
    print(f"Quantidade de tokens antes: {len(tokens_antes)}")
    print(f"Quantidade de tokens depois: {len(tokens_depois)}")
    print(f"Tokens removidos: {tokens_removidos}")
    print(f"Tokens finais: {tokens_depois}")
    print("=" * 50)

# Execução obrigatória do teste
analisar_preprocessamento("MEU sofá!!! chegou quebrado e quero DEVOLVER!!!")
```

```python
# 1. Carregar datasets
df_treino = pd.read_csv("sac_moveis_ac2_treino.csv")
df_teste = pd.read_csv("sac_moveis_ac2_teste.csv")

# 2. Aplicar a esteira de pré-processamento
df_treino["mensagem_limpa"] = df_treino["mensagem"].apply(limpar_e_lemmatizar)
df_teste["mensagem_limpa"] = df_teste["mensagem"].apply(limpar_e_lemmatizar)

# 3. Construir corpus tokenizado
corpus_tokenizado = [msg.split() for msg in df_treino["mensagem_limpa"] if len(msg.split()) > 0]

# 4. Treinar modelo FastText
modelo_fasttext = FastText(
    sentences=corpus_tokenizado,
    vector_size=50,
    window=3,
    min_count=1,
    workers=4,
    sg=1
)

# 5. Função de Mean Pooling
def obter_vetor_frase(frase, modelo):
    palavras = frase.split()
    vetores = []

    for palavra in palavras:
        if palavra in modelo.wv:
            vetores.append(modelo.wv[palavra])

    if len(vetores) == 0:
        return np.zeros(modelo.vector_size)

    return np.mean(vetores, axis=0)

# Gerar vetores numéricos de treino e teste
X_train_vec = np.array([obter_vetor_frase(msg, modelo_fasttext) for msg in df_treino["mensagem_limpa"]])
X_test_vec = np.array([obter_vetor_frase(msg, modelo_fasttext) for msg in df_teste["mensagem_limpa"]])

y_train = df_treino["intencao"]
y_test = df_teste["intencao"]

print(f"Formatos dos conjuntos de vetores: Treino {X_train_vec.shape} | Teste {X_test_vec.shape}")
```

```python
# 1. Treinar modelo de Regressão Logística
modelo_logistico = LogisticRegression(max_iter=1000, random_state=42)
modelo_logistico.fit(X_train_vec, y_train)

# 2. Avaliar modelo no conjunto de teste
y_pred_logistico = modelo_logistico.predict(X_test_vec)
print("Relatório de Classificação (Regressão Logística):")
print(classification_report(y_test, y_pred_logistico))

# 3. Função de classificação com limiar de confiança (Fallback)
def classificar_mensagem(mensagem, modelo, modelo_embedding, limiar=0.50):
    msg_limpa = limpar_e_lemmatizar(mensagem)
    vetor = obter_vetor_frase(msg_limpa, modelo_embedding)

    probas = modelo.predict_proba([vetor])[0]
    max_idx = np.argmax(probas)
    confianca = probas[max_idx]
    intencao_predita = modelo.classes_[max_idx]

    if confianca >= limiar:
        return intencao_predita, confianca
    return "FALLBACK_HUMANO", confianca

# 4. Testes do chatbot em 5 cenários
testes = [
    "quero devolver meu sofá",
    "como faço para realizar a devolução?",
    "cadê meu pedido?",
    "meu pedido nao chego",
    "qual é a previsão do tempo?"
]

print("\n--- TESTE DE PREDICÕES EM TEMPO REAL ---")
for msg in testes:
    intencao, conf = classificar_mensagem(msg, modelo_logistico, modelo_fasttext, limiar=0.50)
    print(f"Mensagem: '{msg}' -> Intenção: {intencao} | Confiança: {conf:.2%}")
```

```python
# 1. Treinar modelo KNN
modelo_knn = KNeighborsClassifier(n_neighbors=3)
modelo_knn.fit(X_train_vec, y_train)

# 2. Obter predições no conjunto de teste
y_pred_knn = modelo_knn.predict(X_test_vec)

# 3. Calcular métricas ponderadas
acc_log = accuracy_score(y_test, y_pred_logistico)
prec_log = precision_score(y_test, y_pred_logistico, average="weighted")
rec_log = recall_score(y_test, y_pred_logistico, average="weighted")
f1_log = f1_score(y_test, y_pred_logistico, average="weighted")

acc_knn = accuracy_score(y_test, y_pred_knn)
prec_knn = precision_score(y_test, y_pred_knn, average="weighted")
rec_knn = recall_score(y_test, y_pred_knn, average="weighted")
f1_knn = f1_score(y_test, y_pred_knn, average="weighted")

# 4. Construir tabela comparativa
df_comparativo = pd.DataFrame({
    "Modelo": ["Regressão Logística", "KNN"],
    "Accuracy": [f"{acc_log:.2%}", f"{acc_knn:.2%}"],
    "Precision": [f"{prec_log:.2%}", f"{prec_knn:.2%}"],
    "Recall": [f"{rec_log:.2%}", f"{rec_knn:.2%}"],
    "F1": [f"{f1_log:.2%}", f"{f1_knn:.2%}"]
})

print("=" * 50)
print("TABELA COMPARATIVA DE DESEMPENHO")
print("=" * 50)
print(df_comparativo.to_string(index=False))
```
