```python
# ================================================================
# LAB 04 - AULA 02 (MLCB)
# DESAFIO NLU PARA AGÊNCIA DE VIAGENS
# ================================================================

# 1. IMPORTAÇÃO DAS BIBLIOTECAS

import pandas as pd

from sklearn.model_selection import train_test_split
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.linear_model import LogisticRegression
from sklearn.metrics import accuracy_score, classification_report


# ================================================================
# 2. CRIAÇÃO DO DATASET
# ================================================================

dados = {
    "frase": [
        # Intenção: comprar_passagem
        "Quero comprar uma passagem para São Paulo",
        "Gostaria de reservar uma passagem para o Rio de Janeiro",
        "Preciso comprar uma passagem de avião",
        "Quero uma passagem para viajar amanhã",
        "Como faço para comprar uma passagem aérea",

        # Intenção: cancelar_reserva
        "Quero cancelar minha reserva",
        "Preciso cancelar a passagem que comprei",
        "Gostaria de cancelar minha viagem",
        "Como posso cancelar minha reserva",
        "Quero desistir da minha passagem",

        # Intenção: falar_atendente
        "Quero falar com um atendente",
        "Preciso conversar com uma pessoa",
        "Pode me transferir para um atendente",
        "Gostaria de falar com alguém da agência",
        "Preciso de ajuda de um funcionário"
    ],

    "intencao": [
        "comprar_passagem",
        "comprar_passagem",
        "comprar_passagem",
        "comprar_passagem",
        "comprar_passagem",

        "cancelar_reserva",
        "cancelar_reserva",
        "cancelar_reserva",
        "cancelar_reserva",
        "cancelar_reserva",

        "falar_atendente",
        "falar_atendente",
        "falar_atendente",
        "falar_atendente",
        "falar_atendente"
    ]
}

df = pd.DataFrame(dados)

print("DATASET:")
display(df)

print("\nQuantidade de frases:", len(df))
print("\nDistribuição das intenções:")
print(df["intencao"].value_counts())


# ================================================================
# 3. SEPARAÇÃO DOS DADOS EM TREINO E TESTE
# ================================================================

X = df["frase"]
y = df["intencao"]

X_treino, X_teste, y_treino, y_teste = train_test_split(
    X,
    y,
    test_size=0.30,
    random_state=42,
    stratify=y
)

print("\nQuantidade de exemplos para treinamento:", len(X_treino))
print("Quantidade de exemplos para teste:", len(X_teste))


# ================================================================
# 4. VETORIZAÇÃO DOS TEXTOS
# ================================================================

vectorizer = TfidfVectorizer()

X_treino_tfidf = vectorizer.fit_transform(X_treino)
X_teste_tfidf = vectorizer.transform(X_teste)

print("\nVetorização realizada com sucesso!")
print("Quantidade de características:", len(vectorizer.get_feature_names_out()))


# ================================================================
# 5. CRIAÇÃO E TREINAMENTO DO MODELO
# ================================================================

modelo = LogisticRegression(max_iter=1000)

modelo.fit(X_treino_tfidf, y_treino)

print("\nModelo treinado com sucesso!")


# ================================================================
# 6. TESTE DO MODELO
# ================================================================

y_pred = modelo.predict(X_teste_tfidf)

acuracia = accuracy_score(y_teste, y_pred)

print("\nRESULTADO DO TESTE")
print("==================")
print("Acurácia:", acuracia)

print("\nRelatório de classificação:")
print(classification_report(y_teste, y_pred, zero_division=0))


# ================================================================
# 7. PREDIÇÕES COM FRASES INÉDITAS
# ================================================================

frases_ineditas = [
    "Gostaria de adquirir um bilhete de avião",
    "Não quero mais minha reserva",
    "Quero conversar com uma pessoa da agência"
]

frases_ineditas_tfidf = vectorizer.transform(frases_ineditas)

previsoes = modelo.predict(frases_ineditas_tfidf)


# ================================================================
# 8. EXIBIÇÃO DAS INTENÇÕES PREVISTAS
# ================================================================

print("\nPREDIÇÕES PARA FRASES INÉDITAS")
print("==============================")

for frase, previsao in zip(frases_ineditas, previsoes):
    print(f"Frase: {frase}")
    print(f"Intenção prevista: {previsao}")
    print("-" * 50)
```
